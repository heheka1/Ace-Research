trying to matplot frequency distribution with scum

```python


import nltk
import matplotlib
import matplotlib.pyplot as plt

with open("SCUMcopy.txt", 'r', encoding='utf-8') as file:
    text=file.read() #ebb: This unpacks the text from the file
    textTokens=text.split()
lowerCaseTokens = [w.lower() for w in textTokens]
print(set(lowerCaseTokens))

scumTextObject = nltk.Text(lowerCaseTokens)

print(f"scum as NLTK Text Object = {scumTextObject}")

scumTextObject.dispersion_plot(["men", "sex", "penis", "pussy", "girl", "woman"])
plt.show()
```
